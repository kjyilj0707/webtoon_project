```python
import requests
import pandas as pd
import time

headers = {"User-Agent": "Mozilla/5.0"}

weeks = ["mon", "tue", "wed", "thu", "fri", "sat", "sun", "dailyPlus"]
ongoing = []

for week in weeks:
    url = f"https://comic.naver.com/api/webtoon/titlelist/weekday?week={week}&order=user"
    res = requests.get(url, headers=headers)
    data = res.json()
    items = data["titleList"]
    for item in items:
        item["week"] = week
        item["status"] = "연재중"
    ongoing.extend(items)
    print(f"{week}: {len(items)}개 수집")
    time.sleep(0.5)

print(f"연재작 총 {len(ongoing)}개")
```

```python
finished = []

for page in range(1, 70):
    url = f"https://comic.naver.com/api/webtoon/titlelist/finished?page={page}&order=UPDATE"
    res = requests.get(url, headers=headers)
    data = res.json()
    items = data["titleList"]
    for item in items:
        item["week"] = None
        item["status"] = "완결"
    finished.extend(items)
    if page % 10 == 0:
        print(f"{page}페이지까지 {len(finished)}개")
    time.sleep(0.5)

print(f"완결작 총 {len(finished)}개")
```

```python
import os

df = pd.DataFrame(ongoing + finished)
print(df.shape)

df.to_csv("webtoon_raw.csv", index=False, encoding="utf-8-sig")

print(os.getcwd())
print(os.path.exists("webtoon_raw.csv"))
```
